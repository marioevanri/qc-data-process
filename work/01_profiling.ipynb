```python
def profile(path):
    df, _ = read_raw(path)
    examples = {c: " | ".join(df[c].dropna().unique()[:6]) for c in df.columns}
    return pd.DataFrame({
        "filled": df.notna().sum(),
        "empty": df.isna().sum(),
        "unique": df.nunique(),
        "examples": pd.Series(examples),
    })
```
